```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
from matplotlib import font_manager as fm
from matplotlib.lines import Line2D

# Font settings 
available_fonts = {f.name for f in fm.fontManager.ttflist}
font_family = "Arial" if "Arial" in available_fonts else "DejaVu Sans"
plt.rcParams["font.family"] = [font_family]
plt.rcParams["mathtext.fontset"] = "custom"
plt.rcParams["mathtext.rm"] = font_family
plt.rcParams["mathtext.it"] = font_family
plt.rcParams["mathtext.bf"] = font_family

# Random seed
rng = np.random.default_rng(42)

# --- Load data ---
file_path = 'general_catalysis_analysis.xlsx'
df = pd.read_excel(file_path)

# --- Extract relevant columns ---
k3HA      = df['k3HA'].astype(float)
k3B       = df['k3B'].astype(float)
k3HA_lo95 = df['k3HA_lo95'].astype(float)
k3HA_hi95 = df['k3HA_hi95'].astype(float)
k3B_lo95  = df['k3B_lo95'].astype(float)
k3B_hi95  = df['k3B_hi95'].astype(float)
ha_pka    = df['ha_pka'].astype(float)
ha_p      = df['ha_p'].astype(float)
ha_q      = df['ha_q'].astype(float)

# --- Compute x-values ---
with np.errstate(divide='ignore', invalid='ignore'):
    log_ratio = np.log10(ha_p / ha_q)
    x_values = ha_pka + log_ratio

# --- Bayesian bootstrap linear regression ---
def bayesian_bootstrap_linreg(x, y, n_draws=5000, rng=None):
    x = np.asarray(x)
    y = np.asarray(y)
    n = len(x)
    W = rng.dirichlet(np.ones(n), size=n_draws)

    wx = W @ x
    wy = W @ y
    Xc = x[None, :] - wx[:, None]
    yc = y[None, :] - wy[:, None]

    cov_xy = np.sum(W * Xc * yc, axis=1)
    var_x = np.sum(W * Xc * Xc, axis=1)

    slope = cov_xy / var_x
    intercept = wy - slope * wx
    return slope, intercept

# --- Plotting function ---
def make_plot(x, y, yerr_lower, yerr_upper, title, y_label, file_name):
    # Classical regression
    slope, intercept, r, _, _ = linregress(x, y)
    r2 = r**2
    y_fit = slope * x + intercept

    # Bayesian bootstrap
    slope_draws, intercept_draws = bayesian_bootstrap_linreg(x, y, rng=rng)
    x_grid = np.linspace(x.min(), x.max(), 200)
    y_pred_draws = intercept_draws[:, None] + slope_draws[:, None] * x_grid[None, :]
    y_low = np.percentile(y_pred_draws, 2.5, axis=0)
    y_med = np.percentile(y_pred_draws, 50, axis=0)
    y_high = np.percentile(y_pred_draws, 97.5, axis=0)

    # Compute credible intervals
    slope_ci = np.percentile(slope_draws, [2.5, 97.5])
    intercept_ci = np.percentile(intercept_draws, [2.5, 97.5])

    # --- Plot ---
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)

    ax.errorbar(x, y, yerr=[yerr_lower, yerr_upper], fmt='o', color='blue',
                ecolor='black', capsize=3, label='Observed data')
    ax.plot(x, y_fit, color='black', linestyle='--', label='Classical fit')
    ax.plot(x_grid, y_med, color='red', lw=2, label='Bayes median fit')
    ax.fill_between(x_grid, y_low, y_high, color='grey', alpha=0.3, label='95% credible interval')

    ax.set_xlabel(r'ha_$\mathit{p}K_\mathrm{a}$ + log$_{10}$(ha_p / ha_q)')
    ax.set_ylabel(y_label)
    ax.set_title(title)
    ax.set_ylim(-6, None)
    p4_custom_legend = [
        Line2D([0],[0], marker='o', color='w', label='Data',       markerfacecolor='blue',       markersize=8),
        Line2D([0],[0], color='black',      lw=2,  label='Linear fit line'),
        Line2D([0],[0], color='black', lw=6, alpha=0.2, label='95% CI')
    ]
    plt.legend(handles=p4_custom_legend,
               bbox_to_anchor=(1.0, 0.25), loc='lower left', frameon=False)

    # Text box with stats
    stats_text = (
        f"Classic Fit:\n"
        f"  y = {slope:.2f}·x + {intercept:.2f}\n"
        f"  R² = {r2:.3f}\n\n"
        f"Bayesian Fit:\n"
        f"  slope = {np.median(slope_draws):.2f}  "
        f"[{slope_ci[0]:.2f}, {slope_ci[1]:.2f}]\n"
        f"  intercept = {np.median(intercept_draws):.2f}  "
        f"[{intercept_ci[0]:.2f}, {intercept_ci[1]:.2f}]\n\n\n\n\n\n"
    )

    ax.text(1.02, 0.55, stats_text, transform=ax.transAxes,
            fontsize=10, va='center',
            bbox=dict(facecolor='white', edgecolor='gray', alpha=0.9))

    plt.savefig(file_name, dpi=300,bbox_inches='tight')
    plt.show()

# --- Plot 1: k3HA ---
mask_ha = (
    (ha_pka < 10) &
    (k3HA > 1e-6) &
    np.isfinite(x_values) &
    np.isfinite(np.log10(k3HA)) &
    (k3HA_lo95 > 0) & (k3HA_hi95 > 0)
)
x_ha = x_values[mask_ha]
y_ha = np.log10(k3HA[mask_ha])
yerr_ha_lower = np.maximum(y_ha - np.log10(k3HA_lo95[mask_ha]), 0)
yerr_ha_upper = np.maximum(np.log10(k3HA_hi95[mask_ha]) - y_ha, 0)

make_plot(
    x=x_ha, y=y_ha,
    yerr_lower=yerr_ha_lower,
    yerr_upper=yerr_ha_upper,
    title=r'Brønsted Plot: k$_{3,HA}$ (ha_$\mathit{p}K_\mathrm{a}$ < 10, k3HA > 1e-5)',
    y_label='log$_{10}$(k$_{3,HA}$)',
    file_name='bronsted_plot_k3HA_bayes_CI.png'
)

# --- Plot 2: k3B ---
mask_b = (
    (ha_pka > 7) &
    (k3B > 1e-5) &
    np.isfinite(x_values) &
    np.isfinite(np.log10(k3B)) &
    (k3B_lo95 > 0) & (k3B_hi95 > 0)
)
x_b = x_values[mask_b]
y_b = np.log10(k3B[mask_b])
yerr_b_lower = np.maximum(y_b - np.log10(k3B_lo95[mask_b]), 0)
yerr_b_upper = np.maximum(np.log10(k3B_hi95[mask_b]) - y_b, 0)

make_plot(
    x=x_b, y=y_b,
    yerr_lower=yerr_b_lower,
    yerr_upper=yerr_b_upper,
    title=r'Brønsted Plot: k$_{3,B}$ (ha_$\mathit{p}K_\mathrm{a}$ > 7, k3B > 1e-5)',
    y_label='log$_{10}$(k$_{3,B}$)',
    file_name='bronsted_plot_k3B_bayes_CI.png'
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
from matplotlib import font_manager as fm

# Font settings 
available_fonts = {f.name for f in fm.fontManager.ttflist}
font_family = "Arial" if "Arial" in available_fonts else "DejaVu Sans"
plt.rcParams["font.family"] = [font_family]
plt.rcParams["mathtext.fontset"] = "custom"
plt.rcParams["mathtext.rm"] = font_family
plt.rcParams["mathtext.it"] = font_family
plt.rcParams["mathtext.bf"] = font_family

# Random seed
rng = np.random.default_rng(42)

# --- Load data ---
file_path = 'general_catalysis_analysis.xlsx'
df = pd.read_excel(file_path)

# --- Extract relevant columns ---
k3HA      = df['k3HA'].astype(float)
k3B       = df['k3B'].astype(float)
k3HA_lo95 = df['k3HA_lo95'].astype(float)
k3HA_hi95 = df['k3HA_hi95'].astype(float)
k3B_lo95  = df['k3B_lo95'].astype(float)
k3B_hi95  = df['k3B_hi95'].astype(float)
pka_corr    = df['pka_corr'].astype(float)
ha_p      = df['ha_p'].astype(float)
ha_q      = df['ha_q'].astype(float)

# --- Compute x-values ---
with np.errstate(divide='ignore', invalid='ignore'):
    log_ratio = np.log10(ha_p / ha_q)
    x_values = pka_corr + log_ratio

# --- Bayesian bootstrap linear regression ---
def bayesian_bootstrap_linreg(x, y, n_draws=5000, rng=None):
    x = np.asarray(x)
    y = np.asarray(y)
    n = len(x)
    W = rng.dirichlet(np.ones(n), size=n_draws)

    wx = W @ x
    wy = W @ y
    Xc = x[None, :] - wx[:, None]
    yc = y[None, :] - wy[:, None]

    cov_xy = np.sum(W * Xc * yc, axis=1)
    var_x = np.sum(W * Xc * Xc, axis=1)

    slope = cov_xy / var_x
    intercept = wy - slope * wx
    return slope, intercept

# --- Plotting function ---
def make_plot(x, y, yerr_lower, yerr_upper, title, y_label, file_name):
    # Classical regression
    slope, intercept, r, _, _ = linregress(x, y)
    r2 = r**2
    y_fit = slope * x + intercept

    # Bayesian bootstrap
    slope_draws, intercept_draws = bayesian_bootstrap_linreg(x, y, rng=rng)
    x_grid = np.linspace(x.min(), x.max(), 200)
    y_pred_draws = intercept_draws[:, None] + slope_draws[:, None] * x_grid[None, :]
    y_low = np.percentile(y_pred_draws, 2.5, axis=0)
    y_med = np.percentile(y_pred_draws, 50, axis=0)
    y_high = np.percentile(y_pred_draws, 97.5, axis=0)

    # Compute credible intervals
    slope_ci = np.percentile(slope_draws, [2.5, 97.5])
    intercept_ci = np.percentile(intercept_draws, [2.5, 97.5])

    # --- Plot ---
    fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)

    ax.errorbar(x, y, yerr=[yerr_lower, yerr_upper], fmt='o', color='blue',
                ecolor='black', capsize=3, label='Observed data')
    ax.plot(x, y_fit, color='black', linestyle='--', label='Classical fit')
    ax.plot(x_grid, y_med, color='red', lw=2, label='Bayes median fit')
    ax.fill_between(x_grid, y_low, y_high, color='grey', alpha=0.3, label='95% credible interval')

    ax.set_xlabel('pKa_corr + log$_{10}$(ha_p / ha_q)')
    ax.set_ylabel(y_label)
    ax.set_title(title)
    ax.set_ylim(-5, None)
    
    p4_custom_legend = [
        Line2D([0],[0], marker='o', color='w', label='Outliers',      markerfacecolor='lightgrey', markersize=8),
        Line2D([0],[0], marker='o', color='w', label='Inliers',       markerfacecolor='blue',       markersize=8),
        Line2D([0],[0], color='black',      lw=2,  label='Linear fit line')
    ]
    plt.legend(handles=p4_custom_legend,
               bbox_to_anchor=(1.0, 0.2), loc='lower left', frameon=False)

    # Text box with stats
    stats_text = (
        f"Classic Fit:\n"
        f"  y = {slope:.2f}·x + {intercept:.2f}\n"
        f"  R² = {r2:.3f}\n\n"
        f"Bayesian Fit:\n"
        f"  slope = {np.median(slope_draws):.2f}  "
        f"[{slope_ci[0]:.2f}, {slope_ci[1]:.2f}]\n"
        f"  intercept = {np.median(intercept_draws):.2f}  "
        f"[{intercept_ci[0]:.2f}, {intercept_ci[1]:.2f}]\n\n\n\n\n\n"
    )

    ax.text(1.02, 0.55, stats_text, transform=ax.transAxes,
            fontsize=10, va='center',
            bbox=dict(facecolor='white', edgecolor='gray', alpha=0.9))

    plt.savefig(file_name, dpi=300,bbox_inches='tight')
    plt.show()

# --- Plot 1: k3HA ---
mask_ha = (
    (ha_pka < 10) &
    (k3HA > 1e-5) &
    np.isfinite(x_values) &
    np.isfinite(np.log10(k3HA)) &
    (k3HA_lo95 > 0) & (k3HA_hi95 > 0)
)
x_ha = x_values[mask_ha]
y_ha = np.log10(k3HA[mask_ha])
yerr_ha_lower = np.maximum(y_ha - np.log10(k3HA_lo95[mask_ha]), 0)
yerr_ha_upper = np.maximum(np.log10(k3HA_hi95[mask_ha]) - y_ha, 0)

make_plot(
    x=x_ha, y=y_ha,
    yerr_lower=yerr_ha_lower,
    yerr_upper=yerr_ha_upper,
    title='Brønsted Plot: k$_{3,HA}$ (pKa < 10, k$_{3,HA}$ > 1e-5)',
    y_label='log10$_{10}$(k$_{3,HA}$)',
    file_name='bronsted_plot_pka_corr_pq_k3HA_bayes_CI.png'
)

# --- Plot 2: k3B ---
mask_b = (
    (ha_pka > 7) &
    (k3B > 1e-5) &
    np.isfinite(x_values) &
    np.isfinite(np.log10(k3B)) &
    (k3B_lo95 > 0) & (k3B_hi95 > 0)
)
x_b = x_values[mask_b]
y_b = np.log10(k3B[mask_b])
yerr_b_lower = np.maximum(y_b - np.log10(k3B_lo95[mask_b]), 0)
yerr_b_upper = np.maximum(np.log10(k3B_hi95[mask_b]) - y_b, 0)

make_plot(
    x=x_b, y=y_b,
    yerr_lower=yerr_b_lower,
    yerr_upper=yerr_b_upper,
    title='Brønsted Plot: k$_{3,B}$ (pKa > 7, k$_{3,B}$ > 1e-5)',
    y_label='log$_{10}$(k$_{3,B}$)',
    file_name='bronsted_plot_pka_corr_pq_k3B_bayes_CI.png'
)
```
