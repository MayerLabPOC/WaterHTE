# Scripts for processing of absorption spectra - Manual version
## Step 1: The first part picks the absorption maximum and determines the extinction coefficient

This version takes an Excel sheet as input, in which the first column is the wavelength, and which is followed by columns containing absorption values (i.e., UV/Vis spectra). The idea of this script is the following: a 96-well plate was charged in a way that each column contains a compound, and each row features the different concentrations of these compounds.
You need a perfectly formatted second Excel input file, which contains your concentrations AND the absorption maxima

The script filters the resulting spectra and discards those with absorption values > 3 due to possible saturation of the detector. It will then generate an overlay plot of the spectra. Finally, it will determine the extinction coefficient for this absorption maximum through a linear regression.

Script version: 2025/07/31

```python
# %% Imports and global style tweaks
%matplotlib inline
import pandas as pd, numpy as np, re
import matplotlib.pyplot as plt
from scipy.stats import linregress

# ---- STYLE SETTINGS ----
plt.rcParams['font.family']         = 'Arial'
plt.rcParams['mathtext.fontset']    = 'dejavusans'    # DejaVu for math symbols
plt.rcParams['axes.unicode_minus']  = False            # ASCII minus
plt.rcParams['font.size']           = 12               # base text size
plt.rcParams['axes.titlesize']      = 12               # subplot titles
plt.rcParams['axes.labelsize']      = 12               # axis labels
plt.rcParams['xtick.labelsize']     = 12               # tick labels
plt.rcParams['ytick.labelsize']     = 12
plt.rcParams['xtick.direction']     = 'in'
plt.rcParams['ytick.direction']     = 'in'
plt.rcParams['xtick.major.size']    = 5
plt.rcParams['ytick.major.size']    = 5
plt.rcParams['xtick.minor.visible'] = False
plt.rcParams['ytick.minor.visible'] = False
# --------------------------------------------------------------------

# %% Load concentration table
concs_df = (
    pd.read_excel('SK147_concentrations.xlsx')
      .set_index('Unnamed: 0')  # rows A–G
)
compounds = list(concs_df.columns)

# %% Load user-provided maxima
max_df = pd.read_excel('SK147_concentrations_maximum.xlsx', index_col=0)
# we expect a row named 'maximum' mapping each compound → wavelength
lambda_max = max_df.loc['maximum'].to_dict()
print("Using provided λₘₐₓ values:")
for comp, lm in lambda_max.items():
    print(f"  {comp}: {lm:.1f} nm")

# %% Load spectra
spectra = pd.read_excel('SK147_pathlength_background.xlsx', index_col='Wavelength')

# %% Build metadata & filter high-abs traces
meta = []
pattern = re.compile(r'\(([A-Z])(\d{2})\)')
for well in spectra.columns:
    m = pattern.search(well)
    if not m:
        continue
    row, col = m.group(1), int(m.group(2))
    if row not in concs_df.index or not (1 <= col <= len(compounds)):
        continue
    comp = compounds[col-1]
    conc = concs_df.at[row, comp]
    if spectra[well].max() <= 3:
        meta.append({'Well': well, 'Compound': comp, 'Concentration': conc})
meta = pd.DataFrame(meta).set_index('Well')

# %% Plot & fit per compound with subplots, saving each as PNG_manual
for comp, lm in lambda_max.items():
    # skip any compounds not in our data
    if comp not in meta['Compound'].values:
        continue

    wells = meta.query("Compound == @comp").index
    concs = meta.loc[wells, 'Concentration'].values
    A_lm = spectra.loc[lm, wells].values

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,4))

    # Left: overlaid spectra with horizontal λₘₐₓ label
    for w, c in zip(wells, concs):
        ax1.plot(spectra.index, spectra[w], label=f"{c:.1e} M")
    ax1.axvline(lm, color='k', ls='--', alpha=0.5)
    ymax = ax1.get_ylim()[1]
    ax1.text(
        lm, ymax*0.95,
        rf"$\lambda_{{\max}}={lm:.1f}\,\mathrm{{nm}}$",
        ha='center', va='bottom'
    )
    ax1.set_title(f"{comp}")
    ax1.set_xlabel("Wavelength (nm)")
    ax1.set_ylabel("Absorbance (AU)")
    ax1.legend(title='[Conc]', loc='upper right', fontsize='small')

    # Right: A vs. [c] & linear fit
    slope, intercept, r, *_ = linregress(concs, A_lm)
    x0 = np.linspace(0, concs.max(), 2)
    ax2.scatter(concs, A_lm)
    ax2.plot(x0, intercept + slope*x0, '--')
    ax2.set_title(
        rf"$\varepsilon = {slope:.2e}\,\mathrm{{L\,mol^{{-1}}\,cm^{{-1}}}}\;(R^2={r**2:.3f})$"
    )
    ax2.set_xlabel("Concentration (M)")
    ax2.set_ylabel(rf"$A(\lambda_{{\max}}={lm:.1f}\,\mathrm{{nm}})$")

    fig.tight_layout()

    # Save PNG with "_manual" suffix
    safe_name = comp.replace(',', '').replace(' ', '_')
    filename = f"{safe_name}_uvvis_fit_manual.png"
    fig.savefig(filename, dpi=300)
    print(f"Saved: {filename}")

    plt.show()

    print(f"{comp}: ε = {slope:.3e} L·mol⁻¹·cm⁻¹  (R² = {r**2:.4f})\n")
```
