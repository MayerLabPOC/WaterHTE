# pH-Buffer Kinetics - Full Analysis - Version 1.2 (2024/08/14)

## Introduction

This scripts takes a series of kinetic measurements that study the dependency of pH-value, buffer-concentration, and nucleophile concentration, and potentially also the buffer type. For each buffer (the script only differentiates between unique pKa values), the data is processed with the following equation: 

(U + 1/eq) * ((k3HA*HA) + (k3B*B) + k2w + k3h*10**(-pH) + k3oh*10**(-(14-pH)))

__Input__

The standard Excel sheet used to set-up experiments. This must include as columns: 'pH', 'pKa', '[Nu]', '[Buffer]', and 'k'. Further, there must be a column named "Nr", which is used to detect the section with the relevant information.

__This script:__
1) extracts the data and puts them into a dataframe

2) Checks how many uinque pKa values are in the dataset (i.e., differnt buffers)

3) Runs a nonlinear least-squares minimization to extract all rate constants and runs an error analaysis

4) Generates a .png for each buffer containing all the key parameters   

__Attention__

x Whereas this script runs the least-squares minimzation on the whole dataset, for most buffer systems only some of the values can actually be determined accurately. Accordingly, the errors for some other rate constants will be massive/the values useless. 

x The script currently also fits the data to extract an equilibrium constant. For large equilibrium constants, the contribution of 1/eq in the term U + 1/eq will become very small, i.e., will not have an effect. For systems with small equilibrium constants, however, eq will be one of the key parameters. However, it is presently not clear how accurate the eq values determine in this way are, as eq is also pH-dependent.

```python
import pandas as pd

# Define the path to the Excel file
file_path = 'SK121.xlsx'

# Load the entire Data sheet to identify the header row dynamically
df_full = pd.read_excel(file_path, sheet_name='Data', header=None)

# Find the row containing the "Nr" entry
header_row_idx = df_full[df_full.apply(lambda row: row.astype(str).str.contains('Nr').any(), axis=1)].index[0]

# Load the data again using the identified header row
df = pd.read_excel(file_path, sheet_name='Data', header=header_row_idx)

# Select only the necessary columns
df_filtered = df[['pH', 'pKa', '[Nu]', '[Buffer]', 'k']].copy()

# Rename the column [Nu] to U
df_filtered.rename(columns={'[Nu]': 'U'}, inplace=True)

# Convert columns to float, coercing errors to NaN
df_filtered = df_filtered.apply(pd.to_numeric, errors='coerce')

# Drop rows with any NaN values
df_filtered.dropna(inplace=True)

# Calculate HA and B
df_filtered['HA'] = (df_filtered['[Buffer]'] * 10**(df_filtered['pKa'] - df_filtered['pH'])) / (1 + 10**(df_filtered['pKa'] - df_filtered['pH']))
df_filtered['B'] = df_filtered['[Buffer]'] - df_filtered['HA']

# Display the dataframe to ensure it loaded and calculated correctly
df_filtered
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import least_squares
from scipy.stats import linregress
import matplotlib.pyplot as plt

# Define the function to calculate k from the model
def calculate_k(params, data):
    pH, U, HA, B = data.T
    k2w, k3HA, k3B, k3h, k3oh, eq = params
    return (U + 1/eq) * ((k3HA*HA) + (k3B*B) + k2w + k3h*10**(-pH) + k3oh*10**(-(14-pH)))

# Define the residual function for least squares optimization
def residuals(params, data, k_observed):
    k_predicted = calculate_k(params, data)
    return k_predicted - k_observed


# Get unique pKa values
pKa_values = df_filtered['pKa'].unique()

# Perform fitting for each unique pKa value
for pKa_value in pKa_values:
    # Filter data for the current pKa value
    subset_df = df_filtered[df_filtered['pKa'] == pKa_value]
    
    # Extract data columns
    data = subset_df[['pH', 'U', 'HA', 'B']].values
    k_observed = subset_df['k'].values

    # Initial guess for fitting parameters
    initial_guess = (1, 1, 1, 1, 1, 1)

    # Define parameter bounds with constraints
    bounds = ([-np.inf, -np.inf, -np.inf, -np.inf, -np.inf, 0],  # Lower bounds
              [np.inf, np.inf, np.inf, np.inf, np.inf, np.inf])  # Upper bounds

    # Perform least squares optimization with tighter convergence criteria and constraints
    result = least_squares(residuals, initial_guess, args=(data, k_observed), ftol=1e-8, xtol=1e-8, bounds=bounds)

    # Extract optimized parameters
    popt = result.x

    # Calculate covariance matrix
    cov_matrix = np.linalg.pinv(result.jac.T @ result.jac)

    # Calculate residual sum of squares
    rss = np.sum(result.fun ** 2)

    # Calculate degrees of freedom
    n_data = len(k_observed)
    n_params = len(popt)
    dof = n_data - n_params

    # Scale covariance matrix by residual sum of squares and degrees of freedom
    cov_matrix_scaled = cov_matrix * rss / dof

    # Calculate standard errors
    perr = np.sqrt(np.diag(cov_matrix_scaled))

    # Calculate residuals
    residual_values = residuals(popt, data, k_observed)

    # Calculate residual sum of squares (RSS)
    rss = np.sum(residual_values ** 2)

    # Calculate degrees of freedom (DOF)
    n_data = len(k_observed)
    n_params = len(popt)
    dof = n_data - n_params

    # Calculate reduced chi-squared (χ²ν)
    chi_squared = rss / dof

    # Calculate root mean squared error (RMSE)
    rmse = np.sqrt(rss / n_data)

    # Calculate coefficient of determination (R²)
    total_sum_of_squares = np.sum((k_observed - np.mean(k_observed))**2)
    r_squared = 1 - (rss / total_sum_of_squares)

    # Print fitting parameters and their standard errors
    print(f"\nResults for pKa = {pKa_value}:")
    print("Optimized Parameters and Standard Errors:")
    print("k2w:", f"{popt[0]:.2e}", "+/-", f"{perr[0]:.2e}")
    print("k3HA:", f"{popt[1]:.2e}", "+/-", f"{perr[1]:.2e}")
    print("k3B:", f"{popt[2]:.2e}", "+/-", f"{perr[2]:.2e}")
    print("k3h:", f"{popt[3]:.2e}", "+/-", f"{perr[3]:.2e}")
    print("k3oh:", f"{popt[4]:.2e}", "+/-", f"{perr[4]:.2e}")
    print("eq:", f"{popt[5]:.2e}", "+/-", f"{perr[5]:.2e}")
    print("\nStatistics:")
    print("Residual Sum of Squares (RSS):", f"{rss:.2e}")
    print("Degrees of Freedom (DOF):", dof)
    print("Reduced Chi-Squared (χ²ν):", f"{chi_squared:.2e}")
    print("Root Mean Squared Error (RMSE):", f"{rmse:.2e}")
    print("Coefficient of Determination (R²):", f"{r_squared:.4f}")

    # Add a new column to subset_df for predicted k values
    subset_df = subset_df.copy()
    subset_df['Predicted_k'] = calculate_k(popt, data)
    subset_df['log_k'] = np.log10(subset_df['k'])
    subset_df['log_Predicted_k'] = np.log10(subset_df['Predicted_k'])
    
    # Plot observed vs. predicted k values
    k_predicted = calculate_k(popt, data)
    plt.scatter(k_observed, k_predicted)
    plt.xlabel('k_observed')
    plt.ylabel('k_predicted')
    plt.title(f'pKa = {pKa_value}: Predicted vs. Observed k')

    # Add linear regression line
    slope, intercept, r_value, p_value, std_err = linregress(k_observed, k_predicted)
    x_vals = np.linspace(min(k_observed), max(k_observed), 100)
    y_vals = slope * x_vals + intercept
    plt.plot(x_vals, y_vals, color='red', label='Linear Regression')

    # Calculate R-squared
    r_squared = 1 - (rss / np.sum((k_observed - np.mean(k_observed)) ** 2))

    # Calculate RMSE
    rmse = np.sqrt(rss / n_data)

    # Annotate fit results
    annotation = f'R² = {r_squared:.2f}\nRMSE = {rmse:.2e}'
    plt.annotate(annotation, xy=(0.05, 0.85), xycoords='axes fraction', fontsize=10)

    # Add result text box
    result_text = f"Optimized Parameters and Standard Errors:\n" \
                f"k2w: {popt[0]:.2e} +/- {perr[0]:.2e}\n" \
                f"k3HA: {popt[1]:.2e} +/- {perr[1]:.2e}\n" \
                f"k3B: {popt[2]:.2e} +/- {perr[2]:.2e}\n" \
                f"k3h: {popt[3]:.2e} +/- {perr[3]:.2e}\n" \
                f"k3oh: {popt[4]:.2e} +/- {perr[4]:.2e}\n" \
                f"eq: {popt[5]:.2e} +/- {perr[5]:.2e}\n" \
                f"RSS: {rss:.2e}\n" \
                f"DOF: {dof}\n" \
                f"χ²ν: {chi_squared:.2e}\n" \
                f"RMSE: {rmse:.2e}\n" \
                f"R²: {r_squared:.4f}"
    plt.text(0.05, 0.1, result_text, transform=plt.gca().transAxes, fontsize=10, verticalalignment='bottom', bbox=dict(facecolor='white', alpha=0.5))

    plt.legend()
    plt.grid(True)
    plt.savefig(f'fit_pKa_{pKa_value}.png')  # Save the plot as an image file
    plt.show()
```
