# Kinetic Analysis: ML-Based Regression Models
Idea: use K_hydration for the carbonyl compounds for parametrization to screen for a pool of ML regression methods in analyzing the data


# Step 1: Importing the data
This cell imports the relevant columns from the dataset. It skips all rows with NN values for kobs, and adds a columns with lg kobs (which we train on later). Finally, it removes the kobs column.


```python
import numpy as np
import pandas as pd

RANDOM_STATE = 42
file_path = "dataset_r2filtered_ml.xlsx"

# Load required inputs; coerce non-numerics to NaN, then drop incomplete rows
usecols = ["lgK_hydration", "ha_pka", "pH_corr", "[nucleophile]", "[ha]tot", "k_obs"]
data = pd.read_excel(file_path, usecols=usecols)
data[usecols] = data[usecols].apply(pd.to_numeric, errors="coerce")
data = data.dropna(subset=usecols).reset_index(drop=True)

# Log-transform target (base 10) with domain check
if (data["k_obs"] <= 0).any():
    raise ValueError("k_obs contains non-positive values; log10 is undefined.")
data["lg_k_obs"] = np.log10(data["k_obs"])

# Model inputs/target
X = data.drop(columns=["k_obs"])
y = data["lg_k_obs"]
target_name = y.name

print(f"Data shape after cleaning: X={X.shape}, y={y.shape}")
X.head()
```

# Step 2: Get an idea of the dataset
This generates a histogram of the data and plots some statistics.


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5.5))
plt.hist(y, bins=100, edgecolor="black", alpha=0.75)
plt.title(f"Distribution of {target_name}")
plt.xlabel(target_name)
plt.ylabel("Frequency")
plt.show()

print(f"{target_name} summary:\n", y.describe())
```

# Step 3: Train/Test Split
This part now performs the train/test split and standardizes the features using the standardscaler

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Guard: ensure target isn't duplicated among features (e.g., after transformations)
X_features = X.drop(columns=[target_name], errors="ignore").copy()

X_train, X_test, y_train, y_test = train_test_split(
    X_features, y, test_size=0.2, random_state=RANDOM_STATE
)

# Standardization for algorithms that assume/benefit from scaled inputs
scaler = StandardScaler().fit(X_train)
X_train_std = scaler.transform(X_train)
X_test_std  = scaler.transform(X_test)

feature_names = X_features.columns.tolist()

print("Shapes:")
print(f"  X_train: {X_train.shape}, X_test: {X_test.shape}")
print(f"  y_train: {y_train.shape}, y_test: {y_test.shape}")
```

# Step 4: ML Training
The following boxes now perform the training of the ML model and test them on the test dataset. For each model, the results are finally plotted.

```python
import os, re, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# Linear / Regularized
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV, ElasticNetCV, BayesianRidge
# Kernel / Neighbors
from sklearn.svm import SVR, LinearSVR
from sklearn.kernel_ridge import KernelRidge
from sklearn.neighbors import KNeighborsRegressor
# Trees / Ensembles
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor, HistGradientBoostingRegressor
# Neural
from sklearn.neural_network import MLPRegressor
# Gaussian Process
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, WhiteKernel, ConstantKernel as C

# Output directory for all figures
PLOTS_DIR = "plots"

# Plot style (set once for all figures)
plt.rcParams.update({
    "figure.dpi": 130,
    "axes.grid": True,
    "grid.linestyle": "--",
    "grid.linewidth": 0.5,
})

def rmse(y_true, y_pred): return float(np.sqrt(mean_squared_error(y_true, y_pred)))
def mae(y_true, y_pred):  return float(mean_absolute_error(y_true, y_pred))

def sanitize(name: str) -> str:
    """Safe filename stem."""
    return re.sub(r"[^A-Za-z0-9_]+", "_", name)

def parity_plot_and_save(name, y_train, y_train_pred, y_test, y_test_pred, target_name, outdir, r2_test, mae_test):
    """Parity plot with y=x guide; saves a transparent PNG."""
    y_train = np.asarray(y_train); y_test = np.asarray(y_test)
    y_all = np.concatenate([y_train, y_test])
    y_min, y_max = float(np.min(y_all)), float(np.max(y_all))

    fig, ax = plt.subplots(figsize=(6, 5.5))
    ax.scatter(y_train, y_train_pred, s=18, alpha=0.7, label="Train")
    ax.scatter(y_test,  y_test_pred,  s=24, alpha=0.7, label="Test")
    ax.plot([y_min, y_max], [y_min, y_max], "k--", lw=1.5, label="y = x")
    ax.set_title(f"{name} — R²(test)={r2_test:.3f}, MAE(test)={mae_test:.3f}")
    ax.set_xlabel(f"Actual {target_name}")
    ax.set_ylabel(f"Predicted {target_name}")
    ax.legend()
    plt.tight_layout()

    os.makedirs(outdir, exist_ok=True)
    fig.savefig(os.path.join(outdir, f"parity_{sanitize(name)}.png"),
                dpi=300, bbox_inches="tight", transparent=True)
    plt.show()

def fit_report_plot(name, model, Xtr, Xte, ytr, yte, target_name, outdir):
    """Fit → metrics → parity plot; returns a metrics dict."""
    model.fit(Xtr, ytr)
    ytr_pred = model.predict(Xtr)
    yte_pred = model.predict(Xte)
    stats = {
        "R2(train)": r2_score(ytr, ytr_pred),
        "R2(test)":  r2_score(yte, yte_pred),
        "MAE(train)": mae(ytr, ytr_pred),
        "MAE(test)":  mae(yte, yte_pred),
        "RMSE(train)": rmse(ytr, ytr_pred),
        "RMSE(test)":  rmse(yte, yte_pred),
    }
    print(f"[{name}] R2(test)={stats['R2(test)']:.3f}, RMSE(test)={stats['RMSE(test)']:.4f}")
    parity_plot_and_save(name, ytr, ytr_pred, yte, yte_pred, target_name, outdir,
                         stats["R2(test)"], stats["MAE(test)"])
    return stats

# Silence occasional GPR linear algebra warnings (expected during kernel optimization)
warnings.filterwarnings("ignore", message=".*divide by zero encountered in matmul.*")
warnings.filterwarnings("ignore", message=".*overflow encountered in matmul.*")
warnings.filterwarnings("ignore", message=".*invalid value encountered in matmul.*")

all_rows = []

def run_family(family_name, models_dict, Xtr, Xte, ytr, yte, target_name, outdir):
    """Run a labeled family of models and collect comparable metrics."""
    for name, model in models_dict.items():
        stats = fit_report_plot(name, model, Xtr, Xte, ytr, yte, target_name, outdir)
        all_rows.append({
            "Family": family_name, "Model": name,
            "R2(test)": stats["R2(test)"], "MAE(test)": stats["MAE(test)"], "RMSE(test)": stats["RMSE(test)"],
            "R2(train)": stats["R2(train)"], "MAE(train)": stats["MAE(train)"], "RMSE(train)": stats["RMSE(train)"],
        })

# Linear / Regularized (scaled)
linear_models = {
    "Linear Regression": LinearRegression(),
    "RidgeCV": RidgeCV(alphas=np.logspace(-6, 6, 25)),
    "LassoCV": LassoCV(alphas=np.logspace(-6, 1, 30), max_iter=20000, random_state=RANDOM_STATE),
    "ElasticNetCV": ElasticNetCV(l1_ratio=np.linspace(0.05, 0.95, 10),
                                 alphas=np.logspace(-6, 1, 30),
                                 max_iter=20000, random_state=RANDOM_STATE),
    "BayesianRidge": BayesianRidge(),
}
run_family("Linear/Regularized", linear_models, X_train_std, X_test_std, y_train, y_test, target_name, PLOTS_DIR)

# Kernel / Neighbors (scaled)
kernel_models = {
    "LinearSVR": LinearSVR(random_state=RANDOM_STATE, max_iter=20000, epsilon=0.1, C=1.0),
    "SVR_RBF": SVR(kernel="rbf", C=1.0, epsilon=0.1, gamma="scale"),
    "SVR_Poly3": SVR(kernel="poly", degree=3, C=1.0, epsilon=0.1, gamma="scale"),
    "KernelRidge(RBF,alpha=1e-1)": KernelRidge(kernel="rbf", alpha=1e-1, gamma=None),
    "kNN(k=5,dist)": KNeighborsRegressor(n_neighbors=5, weights="distance"),
}
run_family("Kernel/Neighbors", kernel_models, X_train_std, X_test_std, y_train, y_test, target_name, PLOTS_DIR)

# Tree Ensembles (unscaled)
tree_models = {
    "RandomForest": RandomForestRegressor(n_estimators=600, max_depth=None, min_samples_leaf=1,
                                          max_features="sqrt", random_state=RANDOM_STATE, n_jobs=-1),
    "GradientBoosting": GradientBoostingRegressor(random_state=RANDOM_STATE),
    "DecisionTree": DecisionTreeRegressor(random_state=RANDOM_STATE),
    "ExtraTrees": ExtraTreesRegressor(n_estimators=600, random_state=RANDOM_STATE, n_jobs=-1),
    "HistGradientBoosting": HistGradientBoostingRegressor(random_state=RANDOM_STATE),
}
run_family("Tree Ensembles", tree_models, X_train, X_test, y_train, y_test, target_name, PLOTS_DIR)

# Neural (scaled)
mlp_models = {
    "MLP(100)": MLPRegressor(hidden_layer_sizes=(100,), activation="relu", solver="adam",
                             alpha=1e-4, learning_rate="adaptive", max_iter=5000,
                             early_stopping=True, random_state=RANDOM_STATE),
}
run_family("Neural", mlp_models, X_train_std, X_test_std, y_train, y_test, target_name, PLOTS_DIR)

# Gaussian Process (scaled)
gpr_kernels = {"GPR_C*(RBF+White)": C(1.0, (1e-3, 1e5)) * (RBF(1.0) + WhiteKernel(1.0))}
for name, ker in gpr_kernels.items():
    gpr = GaussianProcessRegressor(kernel=ker, normalize_y=True,
                                   random_state=RANDOM_STATE, n_restarts_optimizer=2)
    gpr.fit(X_train_std, y_train)
    ytr_pred = gpr.predict(X_train_std)
    yte_pred = gpr.predict(X_test_std)
    stats = {
        "R2(train)": r2_score(y_train, ytr_pred),
        "R2(test)":  r2_score(y_test,  yte_pred),
        "MAE(train)": mae(y_train, ytr_pred),
        "MAE(test)":  mae[y_test,  yte_pred] if False else mae(y_test, yte_pred),  # keep signature; avoids accidental edits
        "RMSE(train)": rmse(y_train, ytr_pred),
        "RMSE(test)":  rmse(y_test,  yte_pred),
        "Kernel(final)": str(gpr.kernel_),
    }
    print(f"[{name}] R2(test)={stats['R2(test)']:.3f}, RMSE(test)={stats['RMSE(test)']:.4f} (kernel→ {stats['Kernel(final)']})")
    parity_plot_and_save(name, y_train, ytr_pred, y_test, yte_pred, target_name, PLOTS_DIR,
                         stats["R2(test)"], stats["MAE(test)"])
    all_rows.append({
        "Family": "Gaussian Process", "Model": name,
        "R2(test)": stats["R2(test)"], "MAE(test)": stats["MAE(test)"], "RMSE(test)": stats["RMSE(test)"],
        "R2(train)": stats["R2(train)"], "MAE(train)": stats["MAE(train)"], "RMSE(train)": stats["RMSE(train)"],
    })

# Consolidated metrics (CSV for downstream tables/figures)
combined = pd.DataFrame(all_rows, columns=[
    "Family", "Model", "R2(test)", "MAE(test)", "RMSE(test)",
    "R2(train)", "MAE(train)", "RMSE(train)"
])
combined.to_csv("results_all_models.csv", index=False)
print("\nSaved: results_all_models.csv")
```

## Automated parameter sweep across carbonyl parameters

```python

# Parameter sweep over carbonyl_parameters.xlsx
# Trains all models for *one carbonyl parameter at a time* (replacing lgK_hydration),
# saves parity plots to parity_plots/ and writes metrics to model_stats_by_parameter.csv
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt, zipfile
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV, ElasticNetCV, BayesianRidge
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

RANDOM_STATE = 42
plots_dir = "parity_plots"
os.makedirs(plots_dir, exist_ok=True)

df_ds = pd.read_excel("dataset_r2filtered_ml.xlsx")
df_params_full = pd.read_excel("carbonyl_parameters.xlsx")

# numeric + log10 target
for c in ["lgK_hydration", "ha_pka", "pH_corr", "[nucleophile]", "[ha]tot", "k_obs"]:
    if c in df_ds.columns:
        df_ds[c] = pd.to_numeric(df_ds[c], errors="coerce")
df = df_ds[df_ds["k_obs"] > 0].copy()
df["lg_k_obs"] = np.log10(df["k_obs"])
if "lgK_hydration" in df.columns:
    df = df.drop(columns=["lgK_hydration"])

join_key = "carbonyl_nr"
param_cols = [c for c in df_params_full.columns if c != join_key]
base_feature_cols = [c for c in ["ha_pka", "pH_corr", "[nucleophile]", "[ha]tot"] if c in df.columns]

models = {
    "LinearRegression": LinearRegression(),
    "RidgeCV": RidgeCV(alphas=np.logspace(-4, 4, 15)),
    "LassoCV": LassoCV(alphas=np.logspace(-4, 1, 16), max_iter=5000, random_state=RANDOM_STATE),
    "ElasticNetCV": ElasticNetCV(l1_ratio=np.linspace(0.1, 0.9, 5),
                                 alphas=np.logspace(-4, 1, 12),
                                 max_iter=5000, random_state=RANDOM_STATE),
    "BayesianRidge": BayesianRidge(),
}

def parity_plot(y_true, y_pred, save_path):
    plt.figure(figsize=(5,5))
    plt.scatter(y_true, y_pred, alpha=0.5, s=8, rasterized=True)
    y_min = float(min(np.min(y_true), np.min(y_pred)))
    y_max = float(max(np.max(y_true), np.max(y_pred)))
    plt.plot([y_min, y_max], [y_min, y_max])
    plt.xlabel("Observed lg_k_obs")
    plt.ylabel("Predicted lg_k_obs")
    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.close()

stats = []
for param in param_cols:
    merged = df.merge(df_params_full[[join_key, param]], on=join_key, how="left")
    feature_cols = base_feature_cols + [param]
    sub = merged.dropna(subset=feature_cols + ["lg_k_obs"]).copy()
    if sub.empty or sub.shape[0] < 10:
        continue
    X = sub[feature_cols].values
    y = sub["lg_k_obs"].values
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=RANDOM_STATE)
    scaler = StandardScaler()
    X_train_s = scaler.fit_transform(X_train)
    X_test_s  = scaler.transform(X_test)
    for model_name, model in models.items():
        model.fit(X_train_s, y_train)
        y_pred_train = model.predict(X_train_s)
        y_pred_test  = model.predict(X_test_s)
        fname = f"{param.replace('/', '_').replace(' ', '_')}__{model_name}.png"
        parity_plot(y_test, y_pred_test, os.path.join(plots_dir, fname))
        stats.append({
            "parameter": param,
            "n_samples": sub.shape[0],
            "n_train": len(y_train),
            "n_test": len(y_test),
            "model": model_name,
            "R2_train": r2_score(y_train, y_pred_train),
            "R2_test": r2_score(y_test, y_pred_test),
            "RMSE_test": mean_squared_error(y_test, y_pred_test, squared=False),
            "MAE_test": mean_absolute_error(y_test, y_pred_test),
        })

pd.DataFrame(stats).to_csv("model_stats_by_parameter.csv", index=False)
with zipfile.ZipFile("parity_plots.zip", "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for fn in os.listdir(plots_dir):
        zf.write(os.path.join(plots_dir, fn), arcname=fn)
print("Done. Files written: model_stats_by_parameter.csv, parity_plots.zip")
```
