```python
import pandas as pd
import numpy as np

# --- Load the main Excel file ---
input_file = 'c1_data_Icorr.xlsx'  
df = pd.read_excel(input_file)

# --- Column indices ---
col_exp_nr      = 1   # Column 2: experiment number
col_buffer      = 5   # Column 6: buffer name
col_pka         = 7   # Column 8: pKa
col_k3ha        = 8   # Column 9: ha_p
col_k3b         = 9   # Column 10: ha_q
col_pH          = 13  # Column 14: pH
col_method      = 27  # Column 28: measurement method

# --- Extract relevant columns ---
df_subset = df.iloc[:, [col_exp_nr, col_buffer, col_pka, col_k3ha, col_k3b, col_pH, col_method]].copy()
df_subset.columns = ['Experiment', 'Buffer', 'pKa', 'ha_p', 'ha_q', 'pH', 'method']

# --- Calculate corrected pKa ---
df_subset['pKa+log(p/q)'] = df_subset['pKa'] + np.log10(df_subset['ha_p'] / df_subset['ha_q'])

# --- Prepare overview table ---
overview_rows = []

for buffer, group in df_subset.groupby('Buffer'):
    exp_nr = group['Experiment'].iloc[0]
    pKa = round(group['pKa'].iloc[0], 2)
    corr_pKa = round(group['pKa+log(p/q)'].iloc[0], 2)
    method = group['method'].iloc[0]
    
    # Extract, round, and sort pH values
    pH_values = sorted(group['pH'].dropna().unique())
    pH_values = [round(pH, 2) for pH in pH_values]

    
    while len(pH_values) < 4:
        pH_values.append(np.nan)

    # Row for the overview table
    row = {
        'Experiment': exp_nr,
        'Buffer': buffer,
        'pKa': pKa,
        'pKa+log(p/q)': corr_pKa,
        'pH1': pH_values[0],
        'pH2': pH_values[1],
        'pH3': pH_values[2],
        'pH4': pH_values[3],
        'method': method
    }
    overview_rows.append(row)

# --- Create the overview DataFrame ---
overview_df = pd.DataFrame(overview_rows)

if np.issubdtype(overview_df['Experiment'].dtype, np.floating):
    overview_df['Experiment'] = overview_df['Experiment'].round(2)

# --- Save the overview table as an Excel file ---
output_file = 'Data_summary.xlsx'
overview_df.to_excel(output_file, index=False, float_format="%.2f")

print(f"Overview successfully saved to: {output_file}")
```
